## 1) Limpieza y generación de features
Ejecuta `scripts/clean_data.py` para crear `data/cleaned.csv`.

```python
import sys, subprocess, os
# Ruta al CSV crudo (ajusta si está en otra ubicación)
RAW_CSV = r'c:\Users\cielo\Downloads\Nueva carpeta\Delitos_Final_2017_2016_2015.csv'
OUT_CLEAN = 'comparative-regression-crime-lima/data/cleaned.csv'
os.makedirs(os.path.dirname(OUT_CLEAN), exist_ok=True)
print('Usando intérprete:', sys.executable)
subprocess.check_call([sys.executable, 'comparative-regression-crime-lima/scripts/clean_data.py', '--input', RAW_CSV, '--output', OUT_CLEAN, '--freq', 'D', '--n-lags', '7'])
print('Limpieza completada — archivo:', OUT_CLEAN)
```

## 3) EDA — mostrar figuras generadas

```python
# Generar figuras (opcional)
subprocess.check_call([sys.executable, 'comparative-regression-crime-lima/scripts/eda.py', '--input', 'comparative-regression-crime-lima/data/cleaned.csv', '--outdir', 'comparative-regression-crime-lima/figures'])
print('Figuras generadas en comparative-regression-crime-lima/figures')
```

## 4) Modelado y tablas comparativas
